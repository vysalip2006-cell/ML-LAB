```python
import pandas as pd
import numpy as np

df = pd.read_excel("Lab Session Data.xlsx", sheet_name="Purchase data")

X = df[['Candies (#)', 'Mangoes (Kg)', 'Milk Packets (#)']].values
y = df[['Payment (Rs)']].values

print("Dimensionality:", X.shape[1])
print("Number of Vectors:", X.shape[0])

rank = np.linalg.matrix_rank(X)
print("Rank:", rank)

cost = np.linalg.pinv(X) @ y
print("Product Costs:")
print(cost)
```

```python
#A4
import math
def minkowski_distance(u,v,p):

    total = 0

    for i in range(len(u)):
        total += abs(u[i]-v[i])**p

    return total**(1/p)
print(minkowski_distance([1,2,3],[4,5,6],1))
```

```python
import matplotlib.pyplot as plt

A = df.select_dtypes(include='number').iloc[0].values
B = df.select_dtypes(include='number').iloc[1].values

distances = []

for p in range(1,11):
    distances.append(
        minkowski_distance(A,B,p)
    )

plt.plot(range(1,11),distances,marker='o')
plt.xlabel("p value")
plt.ylabel("Distance")
plt.title("Minkowski Distance")
plt.show()
```

```python
#A7
import math

A = [1, 2, 3]
B = [4, 5, 6]

def dot_product(u, v):
    total = 0

    for i in range(len(u)):
        total += u[i] * v[i]

    return total


def euclidean_norm(u):
    total = 0

    for i in range(len(u)):
        total += u[i] ** 2

    return math.sqrt(total)


print("Dot Product =", dot_product(A, B))
print("Norm of A =", euclidean_norm(A))
print("Norm of B =", euclidean_norm(B))
```
