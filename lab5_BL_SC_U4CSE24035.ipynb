```python
def bubble_sort(data):

    for i in range(len(data)):

        for j in range(len(data)-i-1):

            if data[j][0] > data[j+1][0]:
                data[j], data[j+1] = data[j+1], data[j]

    return data


data = [(5, 1), (2, 0), (8, 2), (1, 1)]

print(bubble_sort(data))
```

```python
def selection_sort(data):
  n=len(data)
  for i in range(n):
    min_idx=i
    for j in range (i+1, n):
      if data[j][0]<data[min_idx][0]:
        min_idx=j
    data[i],data[min_idx]=data[min_idx],data[i]
  return data
data = [(5, 1), (2, 0), (8, 2), (1, 1)]

print(selection_sort(data))
```

```python
def insertion_sort(data):

    for i in range(1, len(data)):

        key = data[i]
        j = i - 1

        while j >= 0 and data[j][0] > key[0]:

            data[j+1] = data[j]
            j -= 1

        data[j+1] = key

    return data
    data = [(5, 1), (2, 0), (8, 2), (1, 1)]

print(insertion_sort(data))
```

```python
import numpy as np
from sklearn.model_selection import train_test_split # Needed for data splitting
from collections import Counter # Needed for majority voting in predict_one

def predict_one(X_train, y_train, test, k):

    distances = []

    # Calculate Euclidean distance
    for i in range(len(X_train)):

        distance = np.sqrt(
            np.sum(
                (test - X_train[i]) ** 2
            )
        )

        distances.append(
            (distance, y_train[i])
        )

    # Sort according to distance
    distances.sort(
        key=lambda x: x[0]
    )

    # Select K nearest neighbors
    neighbors = distances[:k]

    print("Nearest Neighbors:")
    print(neighbors)

    # Get class labels
    labels = []

    for distance, label in neighbors:
        labels.append(label)

    # Majority voting
    prediction = max(
        set(labels),
        key=labels.count
    )

    return prediction


X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)


predicted_label = predict_one(X_train, y_train, X_test[0], 3)

print("Predicted:", predicted_label)
print("Actual:", y_test[0])
```
